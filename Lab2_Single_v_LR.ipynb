```python
def cost_function(theta_array, x,y,m):
  total_error=0
  for i in range(m):
    total_error+=((theta_array[0]+theta_array[1]*x[i])-y[i])**2
    #total_error+=((theta_array[0]+theta_array[1]*x[1]+theta_array[2]*x[2]+theta_array[3]*x[3])-y[i]))**2
  return (total_error/(2*m))
```

```python
def training(x,y,alpha,iters):
  theta_0,theta_1=0,0
  #theta_0,theta_1,theta_2,theta_3=0,0,0,0
  theta_array=[theta_0,theta_1]
  #theta_array=[theta_0,theta_1,theta_2,theta_3]
  m=x.size
  cost_value=[]
  for i in range(iters):
    theta_array=gradient_descent(theta_array,x,y,m,alpha)
    cost_value.append(cost_function(theta_array,x,y,m))
    y_new=theta_array[0]+theta_array[1]*x
    #y_new=theta_array[0]+theta_array[1]*x[1]+theta_array[2]*x[2]+theta_array[3]*x[3]
    plt.plot(x,y_new,'b')
    plt.scatter(x,y, c="turquoise")
    plt.show()
    x=np.arange(0,len(cost_value),step=1)
    plt.plot(x,cost_value)
    plt.show()
  return theta_array
```

```python
def gradient_descent(theta_array,x,y,m,alpha):
  sum_0,sum_1=0,0
  for i in range(m):
    sum_0+=((theta_array[0]+theta_array[1]*x[i])-y[i])
    sum_1+=x[i]*((theta_array[0]+theta_array[1]*x[i])-y[i])
    # sum_0+=((theta_array[0]+theta_array[1]*x[1]+theta_array[2]*x[2]+theta_array[3]*x[3])-y[i]))
    # sum_1+=x[i]*((theta_array[0]+theta_array[1]*x[1]+theta_array[2]*x[2]+theta_array[3]*x[3])-y[i]))
    # sum_2+=x[i]*((theta_array[0]+theta_array[1]*x[1]+theta_array[2]*x[2]+theta_array[3]*x[3])-y[i])))
    # sum_3+=x[i]*((theta_array[0]+theta_array[1]*x[1]+theta_array[2]*x[2]+theta_array[3]*x[3])-y[i]))
  update_theta0=theta_array[0]-(alpha/m)*sum_0
  update_theta1=theta_array[1]-(alpha/m)*sum_1
  new_theta=[update_theta0,update_theta1]
  return new_theta
```
